```r
# Precisamos instalar o pacote party antes de utilizá-lo
# Similar ao comando !pip isntall do python, utilizaremos o comando install.packages
install.packages("party")
```

```r
# Similar ao comando import do python, que utilizamos para trazer alguma biblioteca
library("party")
```

```r
# Estamos utilizando o ambiente em R para montagem da árvore de decisão (visual)
df = read.csv("https://raw.githubusercontent.com/diogenesjusto/FIAP/master/Gradua%C3%A7%C3%A3o/dados/dados_arvore_titanic.csv")
```

```r
df
```

```r
# Vamos utilizar o comando ctree da biblioteca party para construir uma árvore de decisão
# ele é similar ao sklearn.tree DecisionTreeClassifier
# Para tratarmos a variável Sex não como um texto, e sim como uma categoria
# precisamos transformar a variável em categórica utilizando o comando as.factor

mod = ctree(Survived ~ as.factor(Sex)+as.factor(Pclass)+Age, data=df)

# Apresentação gráfica da árvore de decisão
plot(mod, type='simple')
```
