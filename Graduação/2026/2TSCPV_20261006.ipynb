```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv("https://raw.githubusercontent.com/diogenesjusto/FIAP/refs/heads/master/dados/mtcars.csv")
```

```python
df.plot.scatter(x='wt', y='mpg')
```

```python
# feature eng. - ajustes de escala
# Ajuste de escala utilizando funções do SkLearn
scaler = MinMaxScaler(feature_range = (0,1))
df['wt_01'] = scaler.fit_transform(df[['wt']])
df['mpg_01'] = scaler.fit_transform(df[['mpg']])

df
```

```python
k = KMeans(n_clusters=4, random_state=1).fit(df[["wt_01","mpg_01"]])

df['cluster'] = k.labels_

df.plot.scatter(x='wt', y='mpg', c='cluster', colormap= "viridis")
```
